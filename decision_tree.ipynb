```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)
```

```python
iris = load_iris()
X = iris.data
y = iris.target

print(f"No. of Samples: {X.shape[0]}")
print(f"No. of Features: {X.shape[1]}")
print("\nClass distribution: ")

for i in np.unique(y):
  print(f"{iris.target_names[i]}: {np.sum(y == i)}")
```

```python
def evaluate(y_test, y_pred):
  return [
      round(accuracy_score(y_test, y_pred),4),
      round(precision_score(y_test, y_pred, average='weighted'),4),
      round(recall_score(y_test, y_pred, average='weighted'),4),
      round(f1_score(y_test, y_pred, average='weighted'),4)
  ]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
```

```python
# default hyperparameters
model = DecisionTreeClassifier()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

accuracy, precision, recall, f1 = evaluate(y_test, y_pred)
print(f"Accuracy: {accuracy}")
print(f"Precision: {precision}")
print(f"Recall: {recall}")
print(f"F1 Score: {f1}")
```

```python
criteria = ['gini', 'entropy']
results_criteria = []

for criterion in criteria:
  model = DecisionTreeClassifier(criterion = criterion)
  model.fit(X_train, y_train)
  y_pred = model.predict(X_test)

  accuracy, precision, recall, f1 = evaluate(y_test, y_pred)
  results_criteria.append([
      criterion, accuracy, precision, recall, f1
  ])
results_criteria_df = pd.DataFrame(
    results_criteria,
    columns=['Criterion', 'Accuracy', 'Precision', 'Recall', 'F1 Score']
)
results_criteria_df
```

```python
depths = [1, 2, 3, 5]
results_depths = []

for depth in depths:
  model = DecisionTreeClassifier(max_depth= depth)
  model.fit(X_train, y_train)
  y_pred = model.predict(X_test)

  accuracy, precision, recall, f1 = evaluate(y_test, y_pred)
  results_depths.append([
      depth, accuracy, precision, recall, f1
  ])
results_depths_df = pd.DataFrame(
    results_depths,
    columns=['Depth', 'Accuracy', 'Precision', 'Recall', 'F1 Score']
)
results_depths_df
```

```python
leaf_samples = [1, 2, 5, 10]
results_ls = []

for leaf_sample in leaf_samples:
  model = DecisionTreeClassifier(min_samples_leaf= leaf_sample)
  model.fit(X_train, y_train)
  y_pred = model.predict(X_test)

  accuracy, precision, recall, f1 = evaluate(y_test, y_pred)
  results_ls.append([
      leaf_sample, accuracy, precision, recall, f1
  ])
results_ls_df = pd.DataFrame(
    results_ls,
    columns=['Leaf_Samples', 'Accuracy', 'Precision', 'Recall', 'F1 Score']
)
results_ls_df
```
